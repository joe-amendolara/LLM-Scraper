## 1. The scraper

GPT does not browse the web on its own: we have to download the website's text and hand it over.
That's the job of `fetch_website_contents`, in the [`scraper.py`](scraper.py) file that sits **in the same folder** as this notebook.

<img src="https://raw.githubusercontent.com/joe-amendolara/LLM-Scraper/main/assets/day1_scraper_en.png" alt="Download, clean up, extract text, truncate" width="100%">

| Step | Code | What happens |
| --- | --- | --- |
| 1. Download | `requests.get(url, headers=headers)` | Downloads the page. The `headers` make it look like a Chrome request, because some sites block scripts |
| 2. Clean up | `irrelevant.decompose()` | Removes `script`, `style`, `img` and `input`: code, styling and forms are useless for a summary |
| 3. Extract | `soup.body.get_text(...)` | Keeps only readable text, one element per line |
| 4. Truncate | `[:2_000]` | At most 2,000 characters: less text means faster answers and lower cost |

`scraper.py` is ready to use: no need to copy it here, just import it in the next cell.

## 3. The API key

- `load_dotenv(override=True)` reads the `.env` file and loads the key into memory. `override=True` replaces any older value.
- `os.getenv("OPENAI_API_KEY")` fetches the key by name.
- The three `if` statements are just checks: is the key there? Does it start with `sk-proj-`? Does it have extra spaces?

````python
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    print("No API key found: check your .env file")
elif not api_key.startswith("sk-proj-"):
    print("The key doesn't start with sk-proj-: make sure it's the right one")
elif api_key.strip() != api_key:
    print("The key has spaces at the start or end: remove them from .env")
else:
    print("API key found!")
````

> **Check** — It should print `API key found!`.
> If not, check the `.env` file: exact name, root folder, no spaces or quotes. Then restart the kernel and run from the top again.

## 4. Your first message to GPT

Before the real project, a quick hello to see how simple it is to call a model.

- `OpenAI()` creates the connection. You don't pass the key: it finds it on its own in the `OPENAI_API_KEY` variable.
- `messages` is a **list of messages**. Each one is a dictionary with a `role` (who is speaking) and `content` (what they say).
- `chat.completions.create(...)` sends the conversation to the model chosen with `model`.

<img src="https://raw.githubusercontent.com/joe-amendolara/LLM-Scraper/main/assets/day1_response_en.png" alt="Anatomy of response.choices[0].message.content" width="100%">

OpenAI's response contains a lot of information. The text you want always lives in `response.choices[0].message.content`.
**You will write this line hundreds of times during the course.**

````python
openai = OpenAI()

messages = [{"role": "user", "content": "Hello, GPT! This is my first ever message to you!"}]

response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
response.choices[0].message.content
````

> **Check** — A reply from GPT appears in quotes.
> A quota error (`429`) means you need to add credit on platform.openai.com. An `AuthenticationError` means the key is wrong.

## 6. System prompt and user prompt

<img src="https://raw.githubusercontent.com/joe-amendolara/LLM-Scraper/main/assets/day1_prompts_en.png" alt="System prompt and user prompt become the messages list" width="100%">

Models like GPT expect two kinds of instructions. Picture a brilliant assistant who knows nothing about your situation: you leave them two notes.

| | System prompt | User prompt |
| --- | --- | --- |
| **What it says** | Who you are and how to answer | What to do, with what material |
| **Analogy** | The assistant's job description | The file placed on their desk |
| **When it changes** | Stays the same for every request | Changes on every call |

It's called `user_prompt_prefix` because it's only the **beginning** of the user prompt: the website text gets appended after it.
Triple quotes `"""` let you write text across several lines.

````python
system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

> **Tip** — The last rule (*"Do not wrap the markdown in a code block"*) stops GPT from wrapping its answer in backticks: without it, the output would show raw code instead of formatted text.

## 7. The messages function

`messages_for` takes the website text and returns the list in the format OpenAI expects:
the first message holds the rules (`system`), the second holds the request plus the website (`user`). The `+` joins the two strings.

````python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website},
    ]

messages_for(ed)
````

> **Check** — You see the full list that will be sent: the instructions and, right after the prefix, the website text.
> It's the best way to verify what the model will actually receive.

## 8. The summarize function

`summarize` puts the whole chain together in three steps: it **downloads** the site, **builds** the messages, **calls** GPT and returns the text.
Here we use `gpt-4.1-mini`, a bit more capable than the `nano` model used for the hello.

````python
def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=messages_for(website),
    )
    return response.choices[0].message.content

summarize("https://edwarddonner.com")
````

> **Check** — You see the summary as raw text, with Markdown's `**` asterisks and `#` hashes.

## 9. Formatted output

`Markdown(...)` tells Jupyter *"this text is Markdown"*, and `display(...)` renders it below the cell with real headings and bold text.

````python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))

display_summary("https://edwarddonner.com")
````

> **Check** — The summary appears formatted. The pipeline is complete: URL → scraper → messages → GPT → summary.

## 10. More websites

## 12. Common errors

| Error | Cause | Fix |
| --- | --- | --- |
| `NameError` | You skipped a cell | Run every cell from the top, in order |
| `ModuleNotFoundError` | Wrong kernel or missing library | Choose `.venv`, then `pip install` |
| `No module named 'scraper'` | `scraper.py` isn't next to the notebook | Put it in the `week1/` folder |
| `AuthenticationError` | Wrong or expired key | Check `.env` and restart the kernel |
| Error `429` (quota) | Out of credit | Add credit on platform.openai.com |
| Empty or odd summary | JavaScript site or blocked site | Try another website |

After any change to `.env` or `scraper.py`, **restart the kernel** (the *Restart* button at the top) and run from the beginning: otherwise Python keeps using the old version.
