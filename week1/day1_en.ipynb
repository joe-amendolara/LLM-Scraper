## 1. The scraper

GPT doesn't browse the web on its own: we have to download the website's text and hand it over.
Ed does this with a small utility, `fetch_website_contents`, in the [`scraper.py`](scraper.py) file that sits **in the same folder** as the notebook.

<img src="https://raw.githubusercontent.com/joe-amendolara/LLM-Scraper/main/assets/day1_scraper_en.png" alt="Download, clean up, extract text, truncate" width="100%">

| Step | Code | What happens |
| --- | --- | --- |
| 1. Download | `requests.get(url, headers=headers)` | Downloads the page. The `headers` make it look like a Chrome request, because some sites block scripts |
| 2. Clean up | `irrelevant.decompose()` | Removes `script`, `style`, `img` and `input`: code, styling and forms are useless for a summary |
| 3. Extract | `soup.body.get_text(...)` | Keeps only readable text, one element per line |
| 4. Truncate | `[:2_000]` | At most 2,000 characters: less text means faster answers and lower cost |

The file is ready to use: just import it in the next cell.

## 6. Types of prompts

Models like GPT have been trained to receive instructions in a particular way. They expect:

- a **system prompt** that tells them what task they are performing and what tone they should use;
- a **user prompt**: the conversation starter they should reply to.

<img src="https://raw.githubusercontent.com/joe-amendolara/LLM-Scraper/main/assets/day1_prompts_en.png" alt="System prompt and user prompt become the messages list" width="100%">

| | System prompt | User prompt |
| --- | --- | --- |
| **What it says** | Who you are and how to answer | What to do, with what material |
| **Analogy** | The assistant's job description | The file placed on their desk |
| **When it changes** | Stays the same for every request | Changes on every call |

Ed's system prompt is deliberately **snarky**, to make the result fun. The last sentence asks for Markdown and forbids wrapping it in a code block, otherwise the output would show raw text.

> **Ed says** — later, try changing the last sentence to *Respond in markdown in Spanish.* and see how the answer changes.

````python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

The user prompt is called `user_prompt_prefix` because it's only the **beginning**: the website text gets appended after it. Triple quotes `"""` let you write text across several lines.

````python
# Define our user prompt

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## 7. The messages format

The OpenAI API expects messages in this structure, and many other APIs share it:

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```

Ed makes a very simple call with both roles, using `gpt-4.1-nano`: *"we won't stretch the mighty GPT (yet!)"*.

````python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]

response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
response.choices[0].message.content
````

> **Check** — GPT answers 4. The `system` role set the tone, the `user` role asked the question.

## 8. Building messages with a function

Now Ed builds useful messages for `gpt-4.1-mini` with a function. `messages_for` takes the website text and returns exactly the format above:
the first message holds the rules (`system`), the second the prefix plus the website (`user`). The `+` joins the two strings.

````python
# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

> **Ed says** — try this out, and then try it for a few more websites.

````python
# Try this out, and then try for a few more websites

messages_for(ed)
````

> **Check** — You see the full list that will be sent: the instructions and, after the prefix, the website text. It's the best way to verify what the model will actually receive.

## 9. Calling the OpenAI API

Time to bring it together. `summarize` does three things: it **downloads** the site, **builds** the messages, **calls** `gpt-4.1-mini` and returns the text.

*"The API for OpenAI is very simple! You will get very familiar with this"*, says Ed.

````python
# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
````

````python
summarize("https://edwarddonner.com")
````

> **Check** — You see the summary as raw text, with Markdown's `**` asterisks and `#` hashes.

## 10. Formatted output

`Markdown(...)` tells Jupyter *"this text is Markdown"*, and `display(...)` renders it below the cell with real headings and bold text.

````python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
````

````python
display_summary("https://edwarddonner.com")
````

> **Check** — The summary appears formatted. The pipeline is complete: URL → scraper → messages → GPT → summary.
